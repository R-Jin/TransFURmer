# Dataset Preparation
The SyntheticFur dataset will be used as the dataset. The training data consists of image buffers as input data and high-quality raytraced renders.

- Image Resolution: $1024 \times 1024$
- Format: png
- Directory naming convention: `<scene_name>/<buffer_name><4 digit frame count>.png`
- Frames per Scene: 720
- Buffer names: HighQualityRender, Rasterized, SceneDepth, LitPrimitive, GuideColored, WorldNormal, Mask

## Plan 
What needs to be done to prepare the input data. 

1. Group image buffers that are in the same scene and frame. 
2. Create $16 \times 16$ pixel patches of these image buffers. 
3. Concatenate these to form a multi-channel token that is a $16 \times 16$ token with 14 channels.
